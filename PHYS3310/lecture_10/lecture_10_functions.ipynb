## Why functions

The formula lives in one place and has a name.

```python
print(0.5 * 9.8 * 1 ** 2)
print(0.5 * 9.8 * 2 ** 2)
print(0.5 * 9.8 * 3 ** 2)

def height(t):
    return 0.5 * 9.8 * t ** 2

print(height(1), height(2), height(3))
```

## def

The definition runs nothing; the body runs when the function is called.

```python
def height(t):
    h = 0.5 * 9.8 * t ** 2
    return h

print(height(1))
print(height(3))
a = height(2) + height(4)
print(a)
```

## Several parameters

Arguments are matched to parameters by position.

```python
def kinetic(m, v):
    return 0.5 * m * v ** 2

print(kinetic(2.0, 3.0))
print(kinetic(3.0, 2.0))
```

## Keyword arguments

Matched by name; positional ones come first.

```python
def kinetic(m, v):
    return 0.5 * m * v ** 2

print(kinetic(m=2.0, v=3.0))
print(kinetic(v=3.0, m=2.0))
print(kinetic(2.0, v=3.0))
```

## Default values

Used when the call does not give that argument.

```python
def height(t, g=9.8):
    return 0.5 * g * t ** 2

print(height(3))
print(height(3, 1.6))
print(height(3, g=1.6))
```

## *args

A star before a parameter collects all remaining positional arguments into a tuple.

```python
def total(*values):
    print(values)
    return sum(values)

print(total(1.0, 2.0))
print(total(1.0, 2.0, 3.0, 4.0))
print(total())
```

## **kwargs

Two stars collect all remaining keyword arguments into a dict.

```python
def report(**info):
    print(info)
    for key, value in info.items():
        print(key, "=", value)

report(mass=9.1e-31, charge=-1)
report(name="proton")
```

## * and ** in a call

In a call the stars spread a tuple or a dict over the parameters.

```python
def kinetic(m, v):
    return 0.5 * m * v ** 2

pair = (2.0, 3.0)
print(kinetic(*pair))

settings = {"m": 2.0, "v": 3.0}
print(kinetic(**settings))
```

## Any value can be an argument

The body must make sense for the type it receives.

```python
import numpy as np
def mean(values):
    return sum(values) / len(values)

print(mean([2.1, 2.3, 1.9]))
print(mean((3.0, 4.0)))
print(mean(np.array([1.0, 2.0, 6.0])))

def shout(word):
    return word.upper() + "!"
print(shout("physics"))
```

## return

The call becomes the returned value.

```python
def fall_time(h):
    return (2 * h / 9.8) ** 0.5

t = fall_time(45)
print(t)
print(fall_time(45) * 2)
print(round(fall_time(80), 2))
```

## return several values

They come back as one tuple; unpack it.

```python
def stats(m):
    return min(m), max(m), sum(m) / len(m)

result = stats([2.1, 2.3, 1.9, 2.4])
print(result)
low, high, mean = stats([2.1, 2.3, 1.9, 2.4])
print(low, high, mean)
```

## return ends the function at once

The first return that runs leaves the function.

```python
def first_negative(m):
    for x in m:
        if x < 0:
            return x
    return None

print(first_negative([2.1, -0.5, -1.4]))
print(first_negative([2.1, 2.3]))
```

## An if inside a function

One return per branch.

```python
def state(e):
    if e > 0:
        return "unbound"
    elif e > -1.0:
        return "bound"
    else:
        return "deeply bound"

print(state(0.3), state(-0.5), state(-2.0))
```

## A function that returns True or False

It reads well inside an if.

```python
def is_comment(line):
    return line.startswith("#") or line == ""

print(is_comment("# step energy"))
print(is_comment("0,-0.4996"))

line = "0,-0.4996"
if not is_comment(line):
    print("data:", line)
```

## print is not return

`print` shows; `return` gives back a value.

```python
def show(t):
    print(0.5 * 9.8 * t ** 2)

def give(t):
    return 0.5 * 9.8 * t ** 2

a = show(3)
b = give(3)
print(a, b)
```

## Local names

Names inside the body exist only during the call. The last line is an error on purpose.

```python
def height(t):
    h = 0.5 * 9.8 * t ** 2
    return h

t = 10
print(height(3))
print(t)
print(h)
```

## Reading an outer name

Reading is allowed; assigning makes a new local name. Pass values as parameters instead.

```python
g = 9.8

def height(t):
    return 0.5 * g * t ** 2

def set_moon():
    g = 1.6

set_moon()
print(g, height(3))
```

## Changing an item vs moving the name

The parameter is a second label on the caller's list.

```python
def change_item(m):
    m[0] = 0.0
def change_name(m):
    m = [0.0, 0.0]
a = [2.1, 2.3]
b = [2.1, 2.3]
change_item(a)
change_name(b)
print(a, b)
```

## A loop inside a function

Accumulate inside, return at the end.

```python
def mean(m):
    total = 0
    for x in m:
        total += x
    return total / len(m)

print(mean([2.1, 2.3, 1.9]))
print(mean([1.0, 2.0]))
```

## Loop and if inside a function

The limit is a parameter.

```python
def count_below(m, limit):
    n = 0
    for x in m:
        if x < limit:
            n += 1
    return n

e = [-0.5, -1.2, -0.9, -1.4]
print(count_below(e, -1.0))
print(count_below(e, 0.0))
```

## A function that reads a file

The file name is a parameter.

```python
def read_energies(filename):
    energies = []
    with open(filename) as f:
        for line in f:
            if line.startswith("step"):
                continue
            parts = line.strip().split(",")
            energies.append(float(parts[1]))
    return energies

e = read_energies("data.csv")
print(len(e), min(e), max(e))
```

## Docstring

## Functions calling functions

Each function does one job.

```python
import math

def fall_time(h, g=9.8):
    return math.sqrt(2 * h / g)

def speed_at_ground(h, g=9.8):
    t = fall_time(h, g)
    return g * t

print(fall_time(45))
print(speed_at_ground(45))
```

## Functions from a module

`import math`, `from math import sqrt`, `import numpy as np`.

```python
import math
print(math.sqrt(2), math.pi)

from math import sqrt, pi
print(sqrt(2), pi)

import numpy as np
print(np.sqrt(2))
```

## A function is a value

It can be stored in a name and passed as an argument.

```python
def square(x):
    return x * x

f = square
print(f(3))

def apply_twice(func, x):
    return func(func(x))

print(apply_twice(square, 3))
print(sorted([3, -5, 1], key=square))
```

## lambda

A one-line function without a name, for a short key or formula passed to another function.

```python
square = lambda x: x * x
print(square(3))

print(sorted([3, -5, 1], key=lambda x: x * x))

runs = [("run3", -1.02), ("run1", -0.98), ("run2", -1.10)]
print(sorted(runs, key=lambda r: r[1]))
```
