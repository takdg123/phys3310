## A method and self

`blue.setup(...)` is short for `Ball.setup(blue, ...)`.

```python
class Ball:
    def setup(self, name, m, v):
        self.name = name
        self.m = m
        self.v = v

red = Ball()
Ball.setup(red, "red", 0.5, 3.0)
blue = Ball()
blue.setup("blue", 2.0, 1.5)
print(red.name, blue.name)
```

## A class that holds other objects

```python
class Ball:
    def __init__(self, name, m):
        self.name = name
        self.m = m

class BallList:
    def __init__(self):
        self.balls = []
    def add(self, ball):
        self.balls.append(ball)
    def total_mass(self):
        return sum(b.m for b in self.balls)

box = BallList()
box.add(Ball("red", 0.5))
box.add(Ball("blue", 2.0))
print(len(box.balls), box.total_mass())
```

## __str__

It must return a str.

```python
class Ball:
    def __init__(self, name, m, v):
        self.name = name
        self.m = m
        self.v = v

    def __str__(self):
        return f"{self.name} ball, {self.m} kg, {self.v} m/s"

red = Ball("red", 0.5, 3.0)
print(red)
print(str(red).upper())
```

## Comparing objects

`a < b` calls `a.__lt__(b)`; `sorted` uses it.

```python
class Ball:
    def __init__(self, name, m):
        self.name = name
        self.m = m
    def __lt__(self, other):
        return self.m < other.m
    def __eq__(self, other):
        return self.m == other.m

red = Ball("red", 0.5)
blue = Ball("blue", 2.0)
print(red < blue, red == blue)
for b in sorted([blue, red]):
    print(b.name)
```

## __add__

Return a new object; do not change `self`.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

a = Vector(1.0, 2.0)
b = Vector(3.0, -1.0)
c = a + b
print(c.x, c.y)
print(a.x, a.y)
```

## A value object

The object stores one unit inside, so units cannot be mixed by mistake.

```python
class Length:
    to_meter = {"m": 1.0, "cm": 0.01, "ft": 0.3048}

    def __init__(self, value, unit):
        self.meters = value * self.to_meter[unit]

    def __add__(self, other):
        return Length(self.meters + other.meters, "m")

    def __str__(self):
        return f"{self.meters:.3f} m"

print(Length(3.0, "m") + Length(10.0, "ft"))
```

## Getter and setter

The setter checks every new value.

```python
class Circle:
    def __init__(self, r):
        self.__r = 1.0
        self.set_r(r)
    def get_r(self):
        return self.__r
    def set_r(self, r):
        if r > 0:
            self.__r = r
        else:
            print("radius must be positive")

c = Circle(10)
c.set_r(-5)
print(c.get_r())
```

## Two classes, one copied method

```python
class Circle:
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r ** 2
    def report(self):
        print(">> area:", self.area())
class Square:
    def __init__(self, a):
        self.a = a
    def area(self):
        return self.a ** 2
    def report(self):
        print(">> area:", self.area())

Circle(1.0).report()
Square(2.0).report()
```

## Inheritance

The shared method is written once, in the parent class.

```python
class Shape:
    def report(self):
        print(">> area:", self.area())
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * self.r ** 2
class Square(Shape):
    def __init__(self, a):
        self.a = a
    def area(self):
        return self.a ** 2

Circle(1.0).report()
Square(2.0).report()
```

## Overriding and super()

`super().report()` runs the parent's version.

```python
class Shape:
    def report(self):
        print(">> area:", self.area())

class Square(Shape):
    def __init__(self, a):
        self.a = a
    def area(self):
        return self.a ** 2
    def report(self):
        super().report()
        print("    side:", self.a)

Square(2.0).report()
```

## super().__init__

The parent sets up what every child needs.

```python
class Shape:
    def __init__(self, name):
        self.name = name
    def report(self):
        print(">>", self.name, self.area())

class Square(Shape):
    def __init__(self, a):
        super().__init__("square")
        self.a = a
    def area(self):
        return self.a ** 2

Square(2.0).report()
```

## Composition

The last line is an error on purpose: the list is hidden.

```python
class Measurements:
    def __init__(self):
        self.__values = []
    def add(self, x):
        self.__values.append(x)
    def average(self):
        return sum(self.__values) / len(self.__values)

m = Measurements()
m.add(2.1)
m.add(2.3)
print(m.average())
m[0] = "oops"
```

## A stack

Last in, first out.

```python
class Stack:
    def __init__(self):
        self.__items = []
    def push(self, x):
        self.__items.append(x)
    def pop(self):
        x = self.__items[-1]
        del self.__items[-1]
        return x

s = Stack()
s.push(10)
s.push(20)
s.push(30)
print(s.pop(), s.pop(), s.pop())
```

## A queue

First in, first out.

```python
class Queue:
    def __init__(self):
        self.__items = []
    def enqueue(self, x):
        self.__items.append(x)
    def dequeue(self):
        x = self.__items[0]
        del self.__items[0]
        return x

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.dequeue(), q.dequeue(), q.dequeue())
```
