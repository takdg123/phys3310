## Handling it with if

It works only when you can predict the error.

```python
for text in ["7", "7cm", "-5"]:
    if text.isdigit():
        r = int(text)
        print(r, "->", round(3.14 * r ** 2, 2))
    else:
        print(text, "-> not a whole number")
```

## try / except

Name the error you expect.

```python
for text in ["10", "1.5e3", "ten", "3.0.1"]:
    try:
        x = float(text)
        print(text, "->", x)
    except ValueError:
        print(text, "-> not a number")
```

## Skip the bad lines of a file

`readings.txt` has a word and an empty line in it.

```python
values = []
with open("readings.txt") as f:
    for line in f:
        try:
            values.append(float(line))
        except ValueError:
            print("skipped:", repr(line))
print(values)
```

## A function that asks instead of failing

```python
def is_float(text):
    try:
        float(text)
        return True
    except ValueError:
        return False

m = ["52", "273", "32.1", "x"]
print([is_float(t) for t in m])
print([float(t) for t in m if is_float(t)])
```

## else and finally

"finally" always runs.

```python
for text in ["2.0", "two"]:
    try:
        x = float(text)
    except ValueError:
        print(text, ": except")
    else:
        print(text, ": else,", x * 2)
    finally:
        print(text, ": finally")
```

## finally runs even after return

```python
def read_number(name):
    try:
        f = open(name)
        return int(f.read())
    finally:
        print("finally runs for", name)

print(read_number("number.txt"))
```

## The exception object

`as e` gives the error a name.

```python
try:
    x = int("4two")
except ValueError as e:
    print(type(e))
    print(e)
```

## Several except blocks

One per kind of error.

```python
for name in ["number.txt", "missing.txt", "broken.txt"]:
    try:
        with open(name) as f:
            print(name, int(f.read()))
    except FileNotFoundError:
        print(name, ": no such file")
    except ValueError as e:
        print(name, ":", e)
```

## except Exception comes last

```python
for text in ["4two", "8"]:
    try:
        print(10 / (int(text) - 8))
    except ValueError:
        print(text, ": not a number")
    except Exception as e:
        print(text, ": other error:", e)
```

## raise

Refuse a wrong value with a clear message.

```python
def rectangle_area(w, h):
    if w <= 0 or h <= 0:
        raise ValueError("sides must be positive")
    return w * h

print(rectangle_area(2.0, 3.0))
try:
    print(rectangle_area(0, -1))
except ValueError as e:
    print("error:", e)
```

## raise in a class

No broken object is ever made.

```python
class Circle:
    def __init__(self, r):
        self.set_r(r)
    def set_r(self, r):
        if r <= 0:
            raise ValueError(f"radius must be positive, got {r}")
        self.__r = r

try:
    c = Circle(-10)
except ValueError as e:
    print(e)
```
