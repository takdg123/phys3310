## A comparison is a question

Its value is True or False. `a < b < c` chains. `==` asks, `=` assigns.

```python
t = 5
print(t == 5, t != 5)
print(t < 10, t <= 5, t > 5, t >= 5)
print(0 < t <= 10)
answer = t > 3
print(answer, type(answer))
```

## Comparing carefully

A str never equals a number. Floats: use `math.isclose`. Strings compare letter by letter, so `"10" < "9"`.

```python
import math
print(3 == 3.0, "3" == 3)
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
print(abs(0.1 + 0.2 - 0.3) < 1e-9)
print("apple" < "banana", "Zebra" < "apple")
print("10" < "9")
```

## and

## or

True when at least one side is True.

```python
t = 350
print(t < 0 or t > 300)
print(t < 0 or t > 400)
line = "# step energy"
skip = line == "" or line.startswith("#")
print(skip)
```

## not

Flips the answer. Order: not, then and, then or. Use brackets when in doubt.

```python
t = 5
print(not t > 10)
print(not True, not False)
ok = 0 < t < 10
print(not ok)
print(not (t > 0 and t > 10))
print(not t > 0 or not t > 10)
```

## Every value has a truth

0, 0.0, "", [], None are False; everything else is True.

```python
print(bool(0), bool(0.0), bool(""), bool([]), bool(None))
print(bool(1), bool(-2.5), bool("0"), bool([0]))
m = []
print(not m)
m.append(2.1)
print(not m)
```

## in / not in

An item of a list or set, a key of a dict, a substring of a str. Case-sensitive.

```python
print("H" in ["H", "He", "Li"])
print("error" in "Error: file not found")
print("Error" in "Error: file not found")
mass = {"H": 1.008, "He": 4.003}
print("He" in mass, 4.003 in mass)
print("Fe" not in mass)
print(3 in range(10))
```

## is / is not

`==` compares values, `is` asks for the same object. Use `is` only with None, True, False.

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]
print(a == b, a is b)
print(a == c, a is c)
print(id(a) == id(b), id(a) == id(c))
x = None
print(x is None, x is not None)
```

## Put it together

Name each question; the final rule reads like English. Next: `if keep:`.

```python
t, e_old, e_new, step = 250.0, -1.0032, -1.0033, 12
line = "# step energy"
in_range = 0 <= t <= 300
converged = abs(e_new - e_old) < 1e-3 and step > 10
is_comment = line.startswith("#") or line == ""
keep = in_range and converged and not is_comment
print(in_range, converged, is_comment)
print(keep)
```

## if

`if question:` then an indented block; it runs only when the question is True.

```python
t = 350
if t > 300:
    print("too hot")
    print("stopping")
print("done")
```

## if / else

Exactly one of the two blocks runs.

```python
t = 250
if t > 300:
    label = "too hot"
else:
    label = "ok"
print(label)
label = "hot" if t > 300 else "ok"
print(label)
```

## if / elif / else

Top to bottom; the first True question wins; `else` catches the rest.

```python
e = -0.98
if e > 0:
    state = "unbound"
elif e > -0.5:
    state = "weakly bound"
elif e > -1.0:
    state = "bound"
else:
    state = "deeply bound"
print(state)
```

## Decisions about the data

The file gives the array, a comparison gives the answer, `if` acts on it.

```python
import numpy as np
data = np.loadtxt("data.csv", delimiter=",", skiprows=1)
e = data[:, 1]
print(len(e), e.min(), e.max())
if e.min() < -1.0:
    print("some energy went below -1.0")
if abs(e[-1] - e[-2]) < 1e-3:
    print("converged at the end")
else:
    print("still moving")
```
