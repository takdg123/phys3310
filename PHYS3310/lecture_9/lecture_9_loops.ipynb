## for

`for x in m:` runs the indented block once for each item; `x` is set for you each time.

```python
m = [2.1, 2.3, 1.9]
for x in m:
    print(x, x * 2)
print("done")
```

## Anything with items

A str gives characters, a dict its keys, `.items()` gives key and value.

```python
for ch in "He":
    print(ch)
for k in {"H": 1.008, "He": 4.003}:
    print(k)
for k, v in {"H": 1.008, "He": 4.003}.items():
    print(k, v)
for p in (3.0, 4.0):
    print(p)
for el in {"H", "He", "H"}:
    print(el)
```

## range

`range(4)` is 0, 1, 2, 3. Same rules as a slice: start, stop (not included), step.

```python
for i in range(4):
    print(i)
print(list(range(1, 6)))
print(list(range(0, 10, 2)))
for t in range(4):
    print(t, 0.5 * 9.8 * t ** 2)
```

## An if inside a for

Two indents: the if belongs to the loop, its lines belong to the if.

```python
m = [2.1, 2.3, 1.9, 2.4, 2.2]
largest = m[0]
where = 0
for i in range(len(m)):
    if m[i] > largest:
        largest = m[i]
        where = i
print(largest, where)
print(max(m))
```

## Building a list

Empty list before, `.append()` inside, full list after.

```python
g = 9.8
heights = []
for t in range(5):
    heights.append(0.5 * g * t ** 2)
print(heights)
print(len(heights), heights[-1])
```

## Every line of a file

`for line in f:` walks the file; `continue` skips the header line.

```python
energies = []
with open("data.csv") as f:
    for line in f:
        if line.startswith("step"):
            continue
        parts = line.strip().split(",")
        energies.append(float(parts[1]))
print(len(energies), energies[:3])
print(min(energies))
```

## break and enumerate

`break` leaves the loop at once; `enumerate` gives the number and the item.

```python
with open("log.txt") as f:
    for n, line in enumerate(f, 1):
        if "Error" in line:
            print("first error at line", n)
            print(line.strip())
            break
print("stopped looking")
```

## A loop inside a loop

The outer loop picks a row; the inner loop runs through its values.

```python
import numpy as np
data = np.loadtxt("data.csv", delimiter=",", skiprows=1)
for row in data[:3]:
    for value in row:
        print(value, end="  ")
    print()
for i in range(2):
    for j in range(3):
        print(i, j, end="   ")
```

## List comprehension

A loop that builds a list in one line; an `if` at the end filters.

```python
g = 9.8
heights = [0.5 * g * t ** 2 for t in range(5)]
print(heights)
names = ["run" + str(i) + ".dat" for i in range(1, 4)]
print(names)
cold = [e for e in [2.1, 0.5, 2.4, 0.2] if e < 1]
print(cold)
```

## while

Repeat as long as the question is True. Something in the block must change the answer.

```python
n = 1
while n < 100:
    n = n * 2
    print(n, end=" ")
print()
print("final", n)
```

## while for convergence

Repeat until the error is below a tolerance. Count the steps.

```python
x = 0.0  
target = 1.0  
step = 0

while abs(target - x) > 0.001:
    x = x + (target - x) / 2  
    step += 1

print(f"\nFinal result: Reached ~{x:.5f} after {step} steps!")
```

## Mistake 1

Changing the loop name does not change the list; assign to `m[i]` instead.

```python
m = [2.1, 2.3, 1.9]
for x in m:
    x = x * 2
print(m)
for i in range(len(m)):
    m[i] = m[i] * 2
print(m)
```

## Mistake 2

`range(len(m))` ends at the last index; `m[i + 1]` runs off the end.

```python
m = [2.1, 2.3, 1.9]
for i in range(len(m)):
    print(m[i + 1])
```
