## What you can do with a number

`/` keeps decimals, `//` rounds down, `%` is the remainder, `**` is the power (a fraction gives a root). Powers first, then `* / // %`, then `+ -`; brackets win. Note `-2 ** 2` is -4.

```python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2, 7 // 2, 7 % 2)
print(-7 // 2, -7 % 2)
print(2 ** 10, 2 ** 0.5, 10 ** -3)
print(2 + 3 * 4, (2 + 3) * 4)
print(-2 ** 2, (-2) ** 2)
print(abs(-3.5), round(2.675, 2))
```

## f-strings

An `f` before the quote; anything in `{ }` is computed and put in. `:.2f` gives two decimals, `:.3e` scientific notation.

```python
g = 9.80665
t = 3
print(f"g = {g} m/s^2")
print(f"g = {g:.2f} m/s^2")
print(f"G = {6.674e-11:.3e}")
print(f"after {t} s: {0.5 * g * t**2:.1f} m")
```

## What goes inside the braces

After the colon: decimals `.2f`, scientific `.3e`, thousands `,`, percent `.1%`, a width `12.1f`, alignment `<` `^` `>`, binary `b` and hex `x`, and `{x=}` for debugging.

```python
x = 1234.56789
print(f"{x:.2f}")
print(f"{x:.3e}")
print(f"{x:,.1f}")
print(f"{x:12.1f}|")
print(f"{0.257:.1%}")
print(f"{42:b}  {42:x}  {42:08b}")
print(f"{x=}")
```

## An aligned table

Same width for every value in a column, `>` for numbers, and the decimals line up.

```python
print(f"{'step':>6} {'energy':>10}")
print(f"{0:>6} {-0.4996:>10.4f}")
print(f"{100:>6} {-0.9871:>10.4f}")
print(f"{10000:>6} {-1.0003:>10.4f}")
```

## The other forms you will meet

`%` is the old C style, `.format()` the 2008 style; write f-strings yourself. `\n` and `\t` are special inside quotes. A raw string `r"..."` keeps every backslash, so LaTeX math like `r"$\frac{1}{2} m v^2$"` goes through unchanged to a plot label.

```python
g = 9.80665
print("g = %.2f m/s^2" % g)
print("g = {:.2f} m/s^2".format(g))
print(f"g = {g:.2f} m/s^2")
print("step\tenergy\n0\t-1.0")
print(r"$\frac{1}{2} m v^2$")
```

## Slices, joining, repeating

A third number in a slice is the step; `[::-1]` reverses. `+` joins lists, `*` repeats one. `.index()` and `.count()` search.

```python
m = [2.1, 2.3, 1.9, 2.4, 2.2]
print(m[1:4], m[::2], m[::-1])
print(m + [3.0, 3.1])
print([0] * 5)
print(m.index(1.9), m.count(2.2))
```

## Changing a list

`append` one, `extend` many, `insert` at a position; `pop` takes the last out, `remove` by value, `del` by position; `sort` changes the list in place.

```python
m = [2.1, 2.3, 1.9]
m.append(2.4)
m.extend([2.2, 2.0])
m.insert(0, 1.8)
print(m)
last = m.pop()
print(last, m)
m.remove(2.3)
del m[0]
print(m)
m.sort()
print(m)
```

## Two names, one list

`b = a` does not copy the list. `list(a)` (or `a.copy()`) does. Numbers and strings never have this problem, because they cannot change.

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
c = list(a)
c.append(5)
print(a, c)
```

## How deep does the copy go?

`.copy()` copies only the outer box; the lists inside are shared. `copy.deepcopy()` copies everything inside too.

```python
data = {"step": [0, 1], "energy": [-1.0, -1.0032]}
shallow = data.copy()
shallow["energy"].append(-1.0061)
print(len(data["energy"]))
import copy
deep = copy.deepcopy(data)
deep["energy"].append(-1.0090)
print(len(data["energy"]), len(deep["energy"]))
```

## Working with a dict

`.get()` avoids the KeyError, assigning to a key replaces, `del` removes, `.update()` merges, and `.keys()` `.values()` `.items()` are what a `for` loop will run over.

```python
mass = {"H": 1.008, "He": 4.003}
print(mass.get("C"), mass.get("C", 0.0))
mass["H"] = 1.00794
del mass["He"]
mass.update({"Li": 6.94, "Be": 9.012})
print(mass)
print(list(mass.keys()))
print(list(mass.values()))
print(list(mass.items()))
```

## set

No order, no repeats. `set(list)` drops duplicates; `&` `|` `-` are the math set operations; `in` is very fast. An empty set is `set()`, because `{}` is an empty dict.

```python
runs = ["H", "He", "H", "Li", "He"]
kinds = set(runs)
print(kinds, len(kinds))
a = {"H", "He", "Li"}
b = {"He", "Li", "Be"}
print(a & b)
print(a | b)
print(a - b)
kinds.add("C")
print("C" in kinds, "Fe" in kinds)
```

## np.array

Like a list, but one dtype for all values, packed side by side in memory. Index and slice as with a list. `shape` is the size, as a tuple.

```python
import numpy as np
a = np.array([2.1, 2.3, 1.9, 2.4])
print(a)
print(type(a), a.dtype)
print(a.shape, len(a))
print(a[0], a[-1], a[1:3])
```

## list vs array: math on the whole thing

`m * 2` repeats a list; `a * 2` doubles every value of an array.

```python
m = [2.1, 2.3, 1.9, 2.4]
a = np.array(m)
print(m * 2)
print(a * 2)
print(a + 1)
print(a ** 2)
print(a.mean(), a.max(), a.sum())
print(np.sqrt(a))
```

## list vs array

A list holds any mix and grows; an array holds one dtype, does math at once, and takes a quarter of the memory. The next cell measures the memory row.

```python
import sys
m = [float(i) for i in range(1000)]
a = np.array(m)
print(sys.getsizeof(m) + sum(sys.getsizeof(x) for x in m))
print(a.nbytes)
```

## One dtype for all

NumPy picks the dtype for the whole array. A NumPy int has 64 bits and wraps around silently; a Python int grows.

```python
print(np.array([1, 2, 3]).dtype)
print(np.array([1, 2.0, 3]).dtype)
print(np.array([1, "a"]))
big = np.array([2 ** 62])
print(big * 4)
print(2 ** 62 * 4)
```
